```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("survey.csv")

print(df.shape)
```

```python
# Chart 17: Treatment vs Tech Company

tech_treatment = pd.crosstab(
    df["tech_company"],
    df["treatment"],
    normalize="index"
).mul(100)

tech_treatment.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Treatment by Tech Company")
plt.xlabel("Tech Company")
plt.ylabel("Percentage of Respondents")
plt.xticks(rotation=0)
plt.legend(title="Treatment")
plt.tight_layout()
plt.show()
```

### Why did I choose this chart?
I chose this chart to compare mental health treatment reporting between respondents working in tech companies and non-tech companies.

### Insights
The chart shows the difference in treatment reporting between the two groups. It helps us understand whether treatment reporting varies according to the type of company.

### Business Impact
This insight can help organizations identify whether mental health awareness and support programs may need attention across different workplace environments. However, the chart shows an association and does not prove that working in a tech company directly causes differences in treatment reporting.

### Why did I choose this chart?
